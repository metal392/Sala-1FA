```python
#escrever um programa que recebe as notas do 04 bimestre e imprime a média com o  resultado aprovado, recuperação ou reprovado

#contições: utilizar while
# nota igual ou maior que  7 = aprovado
#nota igual ou maior que 5 = recuperação
#nota menor que 5 = reprovado

bimestre = 1
total_notas = 0

while bimestre <= 4:
  try:
    nota = float(input(f"Digite a nota do {bimestre}º bimestre: "))
    if 0 <= nota <= 10:
      total_notas += nota
      bimestre += 1
    else:
      print("A nota deve estar entre 0 e 10.")
  except ValueError:
    print("Entrada inválida. Por favor, digite um número.")

if bimestre > 1:
  media = total_notas / (bimestre -1)
  print(f"\nMédia final: {media:.2f}")

  if media >= 7:
    print("Resultado: Aprovado")
  elif media >= 5:
    print("Resultado: Recuperação")
  else:
    print("Resultado: Reprovado")
else:
  print("Nenhuma nota foi inserida para calcular a média.")
```
